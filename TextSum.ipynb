```python
# kolnpy: 한글 자연어처리 라이브러리
# sklearn: 머신러닝과 텍스트 분석 라이브러리
# numpy: 숫자 계산과 배열(행렬) 처리 라이브러리
from konlpy.tag import Okt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# 형태소 분석기(문장에서 명사만 추출하는 도구)
okt = Okt()

# 토큰화(문장을 명사로 바꾸는 작업)
# ex) 나는 밥을 먹었다  ->  나 밥
def tokenize(sent):
    return " ".join(okt.nouns(sent))

# 핵심 요약 함수(입력된 글을 3개 문장으로 요약)
def summarize(text, top_n=3):
    # 마침표를 기준으로 문장 분리(+ 짧은 문장은 제외)
    sentences = [s.strip() for s in text.split(".") if len(s.strip()) > 5]

    # 문장 수가 적으면 그대로 반환
    if len(sentences) <= top_n:
        return sentences


    tokenized = [tokenize(s) for s in sentences]

    # 문장을 TF-IDF 방식으로 수치화
    # TF-IDF 방식: 특정 문장에서 자주 쓰이는 단어에 가중치 부여
    vectorizer = TfidfVectorizer()
    tfidf = vectorizer.fit_transform(tokenized)

    # 문장 간 유사도 계산(코사인 유사도 방식 이용)
    sim_matrix = cosine_similarity(tfidf, tfidf)

    # TextRank 방식을 활용하여 문장마다 유사도를 더한 후 중요도 계산
    # TextRank 방식: 중요도 점수 = 유사도 합
    scores = np.sum(sim_matrix, axis=1)

    # 중요도가 높은 문장 top_n 선택 & 글 순서대로 정렬 후 최종 요약결과 반환
    top_idx = sorted(scores.argsort()[-top_n:])
    summary = [sentences[i] for i in top_idx]

    return summary
```

```python
if __name__ == "__main__":
    # 사용자에게 텍스트 입력받기
    print("한글 텍스트를 입력하세요 (한 줄로 입력 후 Enter):")
    text = input()

    # 요약함수 실행
    result = summarize(text, top_n=3)

    # 요약결과 출력
    print("\n📝 요약 결과:")
    for i, s in enumerate(result, 1):
        print(f"{i}. {s}")
```
